```python
import sqlite3
```

```python
connection=sqlite3.connect("ex.db")
```

```python
cursor=connection.cursor()
```

```python
#table
cursor.execute('''
Create Table If Not Exists employees(
id Integer Primary Key,
name Text Not Null,
age Integer,
department text)
''')

connection.commit()
```

```python
#insert 
cursor.execute('''
INSERT INTO employees(name, age, department)
values("Kritish", 32, "Electrical")''')
cursor.execute('''
INSERT INTO employees(name, age, department)
values("Bipul", 35, "Electrical")''')
cursor.execute('''
INSERT INTO employees(name, age, department)
values("Adarsh", 37, "Computer")''')

connection.commit()
```

```python
#query
cursor.execute('Select * from employees')
rows=cursor.fetchall()
for r in rows:
    print(r)
```

```python
cursor.execute(''' 
UPDATE employees
Set age=40
Where name="Bipul"''')
connection.commit()
```
