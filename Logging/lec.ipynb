```python
import logging

logging.basicConfig(
    filename='app.log',
    filemode="w",
    level=logging.DEBUG,
    format='%(asctime)s-%(name)s-%(levelname)s-%(message)s',
    datefmt='%Y-%m-%d %H:%M:%S'
)

logging.debug("This is an debug msg")
logging.info("Info msg")
logging.warning("Warning msg")
logging.error("error msg")
logging.critical("critical message")
```

```python
import logging
l1=logging.getLogger("mod1")
l1.setLevel(logging.DEBUG)

l2=logging.getLogger("mod2")
l2.setLevel(logging.WARNING)
logging.basicConfig(
    filename='app.log',
    filemode="w",
    level=logging.DEBUG,
    format='%(asctime)s-%(name)s-%(levelname)s-%(message)s',
    datefmt='%Y-%m-%d %H:%M:%S'
)
```

```python
l1.debug("Debug")
l2.warning("warn")
```
