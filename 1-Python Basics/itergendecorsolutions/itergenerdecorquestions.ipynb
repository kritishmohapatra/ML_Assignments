### Assignment 15: Generator Pipeline

Write three generator functions: `integers` that yields integers from 1 to 10, `doubles` that yields each integer doubled, and `negatives` that yields the negative of each doubled value. Chain these generators to create a pipeline that produces the negative doubled values of integers from 1 to 10.

```python
class Countdown:
    def __init__(self, start):
        self.current=start
    def __iter__(self):
        return self
    def __next__(self):
        if self.current<=0:
            raise StopIteration
        else:
            self.current-=1
            return self.current
for n in Countdown(5):
    print(n)

class Myrange:
    def __init__(self, st, en):
        self.st=st
        self.en=en
    def __iter__(self):
        return self
    def __next__(self):
        if self.st>=self.en:
            raise StopIteration
        else:
            self.st+=1
            return self.st-1
for n in Myrange(1,6):
    print(n)
def fibo(n):
    a,b=0,1
    for _ in range(n):
        yield a
        a, b=b,a+b
for n in fibo(10):
    print(n)

sq=(x**2 for x in range(1, 11))
for s in sq:
    print(s)
def ev(l):
    for i in range(l+1):
        if i%2==0:
            yield i
def sq(n):
    for num in n:
        yield num**2
e=ev(10)
s=sq(e)
for n in s:
    print(n)

import time
def time_it(func):
    def wrapper(*args, **kwargs):
        st=time.time()
        re=func(*args, **kwargs)
        en=time.time()
        print(f"Execution time {en-st}")
        return re
    return wrapper
@time_it
def fact(n):
    if n==0:
        return 1
    else:
        return n*fact(n-1)
print(fact(10))
def repeat(n):
    def decorator(func):
        def wrapper(*args, **kwargs):
            for _ in range(n):
                func(*args, **kwargs)
        return wrapper
    return decorator

@repeat(3)
def print_message(message):
    print(message)
print_message("Hello, World!")
def upper(func):
    def wr(*args, **kwargs):
        res=func(*args, **kwargs)
        return res.upper()
    return wr
def exclaim(fun):
    def wr(*args, **kwargs):
        res=fun(*args, **kwargs)
        return res+"!"
    return wr 
@upper
@exclaim
def greet(name):
    return "Hello "+name


print(greet(" Kritish"))
def sin(cls):
    inst={}
    def gett_in(*args, **kwargs):
        if cls not in inst:
            inst[cls]=cls(*args, **kwargs)
        return inst[cls]
    return gett_in
@sin
class Data:
    def __init__(self):
        print("Connection")
db1=Data()
d2=Data()
print(db1 is d2)

def inte():
    for i in range(1, 11):
        yield i 
def do(n):
    for num in n:
        yield num*2
def neg(n):
    for num in n:
        yield -num 
int_gen=inte()
doub=do(int_gen)
nege=neg(doub)
for i in nege:
    print(i)
class Infi:
    def __init__(self, start):
        self.start=start
    def  __iter__(self):
        return self
    def __next__(self):
        self.start+=1
        return self.start
s=Infi(0)
for i in range(10):
    print(next(s))  
def selfd(num, d):
    for n in num:
        try:
            yield n/d 
        except ZeroDivisionError:
            yield "Error: Zero divion error"
numbers = [10, 20, 30, 40]
for result in selfd(numbers, 0):
    print(result)      
def count(start):
    cur=start
    while True:
        yield cur 
        cur+=1
c=count(0)
for _ in range(10):
    print(next(c))
def up(cls):
    class W(cls):
        def __init__(self, *args, **kwargs):
            super().__init__(*args, **kwargs)
            self.data=self.data.upper()
    return W

@up
class R:
    def __init__(self, data):
        self.data=data
        self.index=len(data)
    def __iter__(self):
        return self 
    def __next__(self):
        if self.index==0:
            raise StopIteration
        self.index-=1
        return self.data[self.index]
    
for char in R("hello"):
    print(char)

def open_file(file_name, mode):
    def deco(func):
        def wrap(*args, **kwargs):
            with open(file_name, mode) as file:
                return func(file, *args, **kwargs)
        return wrap
    return deco

@open_file("sample.txt","w")
def writee(file, txt):
    file.write(txt)

writee('Hello, World!')
```
